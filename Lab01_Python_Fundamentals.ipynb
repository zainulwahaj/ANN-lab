### Task 6 (a)

> Take a marks value (0-100) stored in a variable and use `if` / `elif` / `else` to print the grade (A, B, C, D, or F) based on standard grading criteria.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 6 (a)
marks = 87        # a value between 0 and 100

# Standard grading criteria:
#   90-100 -> A,  80-89 -> B,  70-79 -> C,  60-69 -> D,  below 60 -> F
if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
elif marks >= 60:
    grade = "D"
else:
    grade = "F"

print("Marks :", marks)
print("Grade :", grade)
```

### Task 6 (c)

> Write nested `if` statements to check whether a given year is a leap year.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 6 (c)
year = 2024

# A year is a leap year if it is divisible by 4,
# except century years, which must also be divisible by 400.
if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(year, "is a LEAP year")
        else:
            print(year, "is NOT a leap year (century year not divisible by 400)")
    else:
        print(year, "is a LEAP year")
else:
    print(year, "is NOT a leap year")
```

## Task 7: Loops with Break and Continue

**Topics Covered:** Flow Control (`for` / `while` / `break` / `continue`)

### Task 7 (a)

> Use a `for` loop to print all numbers from 1 to 30, but skip every multiple of 4 using `continue`.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 7 (a)
kept = []

for n in range(1, 31):
    if n % 4 == 0:
        continue          # skip multiples of 4 and jump to the next iteration
    print(n, end=" ")
    kept.append(n)

print("\n\nNumbers printed:", len(kept), "| skipped:", 30 - len(kept))
```

### Task 7 (b)

> Use a `for` loop to print numbers from 1 to 50, and stop the loop as soon as a number greater than 20 is found, using `break`.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 7 (b)
for n in range(1, 51):
    if n > 20:
        print("\n\nFound", n, "which is greater than 20 -> breaking out of the loop")
        break             # leaves the loop immediately
    print(n, end=" ")
```

### Task 7 (c)

> Use a `while` loop to print the multiplication table (1 to 10) of a number provided in a variable.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 7 (c)
number = 7
i = 1

print(f"Multiplication table of {number}:\n")
while i <= 10:
    print(f" {number} x {i:>2} = {number * i}")
    i += 1                # without this the loop would never end
```

### Task 7 (d)

> Write a loop that calculates and prints the sum of all even numbers between 1 and 100.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 7 (d)
total = 0

for n in range(1, 101):
    if n % 2 == 0:
        total += n

print("Sum of all even numbers between 1 and 100:", total)

# Cross-check with Python's built-in sum()
print("Verified with sum():", sum(n for n in range(1, 101) if n % 2 == 0))
```

### Task 8 (d)

> Write a function that takes a list as a parameter, adds one item to it inside the function, and prints the list before and after the function call to demonstrate how lists are passed by reference.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 8 (d)
def add_item(items):
    """Append an item to the list that was passed in."""
    print("   (inside function) id of parameter :", id(items))
    items.append("Deep Learning")

subjects = ["ANN", "NLP", "ML"]

print("Before function call :", subjects)
print("   (outside) id of subjects        :", id(subjects))

add_item(subjects)

print("After  function call :", subjects)

# The list changed even though the function returned nothing, because the
# parameter 'items' and the variable 'subjects' point to the SAME list object
# (identical id above). Lists are mutable and passed by reference, so append()
# modifies the original object rather than a copy.
```

## Task 9: Classes and Inheritance

**Topics Covered:** Classes (OOP Basics)

### Task 9 (a)

> Create a class `Course` with an `__init__` method that stores `course_name` and `credit_hours`, and a method `display_info()` that prints these details.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 9 (a)
class Course:
    def __init__(self, course_name, credit_hours):
        self.course_name = course_name
        self.credit_hours = credit_hours

    def display_info(self):
        print(f"Course Name : {self.course_name}")
        print(f"Credit Hours: {self.credit_hours}")

print("Class 'Course' defined ->", Course)
```

### Task 9 (b)

> Create two objects of the `Course` class with different data and call `display_info()` on both.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 9 (b)
course1 = Course("Artificial Neural Networks", 3)
course2 = Course("Natural Language Processing", 3)

print("--- Course 1 ---")
course1.display_info()

print("\n--- Course 2 ---")
course2.display_info()
```

### Task 9 (c)

> Create a class `LabCourse` that inherits from `Course` and adds an extra attribute `lab_hours`, using `super().__init__()` to reuse the parent constructor.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 9 (c)
class LabCourse(Course):
    def __init__(self, course_name, credit_hours, lab_hours):
        super().__init__(course_name, credit_hours)   # reuse the parent constructor
        self.lab_hours = lab_hours

lab1 = LabCourse("ANN Lab", 1, 3)

print("Is LabCourse a subclass of Course?", issubclass(LabCourse, Course))
print("Is lab1 an instance of Course?    ", isinstance(lab1, Course))
print()
print("Inherited attributes ->", lab1.course_name, "|", lab1.credit_hours)
print("New attribute        ->", lab1.lab_hours)
```

### Task 9 (d)

> Override the `display_info()` method in `LabCourse` so that it prints all the parent's details plus the `lab_hours`, and demonstrate it by creating one `LabCourse` object.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 9 (d)
class LabCourse(Course):
    def __init__(self, course_name, credit_hours, lab_hours):
        super().__init__(course_name, credit_hours)
        self.lab_hours = lab_hours

    def display_info(self):
        super().display_info()        # print the parent's details first
        print(f"Lab Hours   : {self.lab_hours}")

lab_course = LabCourse("Artificial Neural Networks Lab", 1, 3)

print("--- LabCourse (overridden display_info) ---")
lab_course.display_info()

print("\n--- Parent Course for comparison ---")
course1.display_info()
```
