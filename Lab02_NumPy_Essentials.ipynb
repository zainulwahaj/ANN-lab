## Task 1: Vectors

## Task 3: Addition & Subtraction

**Topic:** Element-wise matrix arithmetic

### Task 3 (a)

> Create two 3x3 matrices representing the marks of 3 students in 3 subjects for Term 1 and Term 2.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 3 (a)
# Rows    = 3 students (Ali, Sara, Bilal)
# Columns = 3 subjects (Maths, Physics, Programming)
students = ["Ali", "Sara", "Bilal"]
subjects = ["Maths", "Physics", "Programming"]

term1 = np.array([
    [72, 65, 80],
    [88, 91, 79],
    [55, 60, 68]
])

term2 = np.array([
    [78, 70, 85],
    [85, 94, 90],
    [70, 66, 62]
])

print("Term 1 marks:")
print(term1)
print("\nTerm 2 marks:")
print(term2)
```

### Task 3 (b)

> Calculate the total combined marks (Term 1 + Term 2) for each student in each subject.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 3 (b)
total_marks = term1 + term2

print("Combined marks (Term 1 + Term 2):")
print(total_marks)

# The + operator works element by element, position against matching position
print("\nChecking one cell by hand:")
print(f"  {students[0]} in {subjects[0]}: {term1[0, 0]} + {term2[0, 0]} = {total_marks[0, 0]}")
```

## Task 4: Dot Product

**Topic:** `np.dot()` and matrix multiplication

### Task 4 (a)

> Create a vector of 4 item prices and a vector of the quantities purchased for each item.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 4 (a)
items = ["Notebook", "Pen", "Backpack", "Calculator"]

prices = np.array([250, 45, 1800, 950])      # price per unit
quantities = np.array([3, 10, 1, 2])         # how many of each were bought

print(f"{'ITEM':<12}{'PRICE':>8}{'QTY':>6}")
print("-" * 26)
for name, price, qty in zip(items, prices, quantities):
    print(f"{name:<12}{price:>8}{qty:>6}")
```

### Task 4 (b)

> Calculate the total bill using the dot product (not by multiplying and adding manually).

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 4 (b)
total_bill = np.dot(prices, quantities)

print("Prices     :", prices)
print("Quantities :", quantities)
print("\nTotal bill :", total_bill)

# The dot product multiplies matching positions and adds the results:
# (250*3) + (45*10) + (1800*1) + (950*2)
print("\nWhat the dot product did internally:")
print("  ", " + ".join(f"({p}x{q})" for p, q in zip(prices, quantities)), "=", total_bill)
```

### Task 4 (c)

> Create a 2x4 matrix representing quantities purchased by 2 different customers, then calculate the total bill for both customers using a single dot product operation.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 4 (c)
customers = ["Customer 1", "Customer 2"]

# Each ROW is one customer, each COLUMN is one item
customer_quantities = np.array([
    [3, 10, 1, 2],
    [5,  2, 2, 1]
])

print("Quantities matrix (2 customers x 4 items):")
print(customer_quantities)
print("Shape:", customer_quantities.shape)

# One dot product handles both customers at once
bills = np.dot(customer_quantities, prices)

print("\nBills:", bills)
for name, amount in zip(customers, bills):
    print(f"  {name}: {amount}")
```

### Task 4 (d)

> Explain (in a comment in your code) why the shapes of the two arrays had to match a specific way for the dot product to work.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 4 (d)
print("customer_quantities shape :", customer_quantities.shape)
print("prices shape              :", prices.shape)
print("result shape              :", bills.shape)

# WHY THE SHAPES HAVE TO LINE UP
# ------------------------------
# A dot product pairs each column of the left array with each row of the right
# array, so the INNER dimensions must be equal:
#
#       (2, 4)  .  (4,)   ->   (2,)
#           ^^^^^^^^
#           these must match
#
# We have 4 items, so the quantities matrix has 4 columns and the prices vector
# has 4 entries. Each of those 4 columns lines up with the price of the same
# item, which is exactly the pairing we want. The OUTER dimensions survive into
# the answer: 2 customers in, 2 totals out.
#
# Reversing the order breaks that rule, because 4 would then face 2:

try:
    np.dot(prices, customer_quantities)
except ValueError as err:
    print("\nnp.dot(prices, customer_quantities) fails ->")
    print("   ValueError:", err)
```

### Task 5 (d)

> Try reshaping the array into a shape that is not possible (e.g., 5x3) and note down the error message you get.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 5 (d)
# 5 x 3 = 15, but the array only holds 12 values, so NumPy refuses.
try:
    impossible = hours.reshape(5, 3)
except ValueError as err:
    print("Attempting hours.reshape(5, 3) ->")
    print("   ValueError:", err)

# Noted: NumPy raises
#     ValueError: cannot reshape array of size 12 into shape (5,3)
# A reshape is only allowed when the new dimensions multiply to exactly the
# number of elements already in the array.

print("\nValid shapes for 12 elements:")
for rows in range(1, 13):
    if 12 % rows == 0:
        print(f"   {rows} x {12 // rows}")
```

### Task 8 (d)

> Without changing the seed, run your random decimal generation code (part 1) twice and compare, explain in a comment why both outputs are identical.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 8 (d)
# Re-seeding with the SAME value before each call
np.random.seed(58)
first_run = np.random.rand(3, 3)

np.random.seed(58)
second_run = np.random.rand(3, 3)

print("First run:")
print(first_run)
print("\nSecond run:")
print(second_run)
print("\nAre the two runs identical?", np.array_equal(first_run, second_run))

# WHY THEY COME OUT IDENTICAL
# ---------------------------
# NumPy's numbers are pseudo-random, not truly random. They come from a formula
# whose output depends entirely on an internal state. seed(58) sets that state to
# a fixed starting point, so the same formula runs from the same place and must
# produce the same sequence. That is what makes experiments reproducible.
#
# IMPORTANT DETAIL: it is the RE-SEEDING that makes them match, not merely
# leaving the seed alone. Each call to rand() advances the internal state, so
# calling it twice in a row WITHOUT re-seeding gives different numbers:

np.random.seed(58)
call_one = np.random.rand(3, 3)
call_two = np.random.rand(3, 3)        # no re-seed in between

print("\nTwo back-to-back calls without re-seeding are identical?",
      np.array_equal(call_one, call_two))
```

## Task 9: Transpose

**Topic:** `.T` and shape compatibility

### Task 9 (a)

> Create a 3x5 matrix representing 3 employees' performance scores across 5 months.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 9 (a)
employees = ["Ayesha", "Hamza", "Zara"]
months = ["Jan", "Feb", "Mar", "Apr", "May"]

performance = np.array([
    [78, 82, 85, 80, 88],
    [65, 70, 68, 75, 79],
    [90, 87, 92, 95, 91]
])

print("Performance matrix (3 employees x 5 months):")
print(performance)
print("\nShape:", performance.shape)

for row, name in enumerate(employees):
    print(f"  {name:<8}: {performance[row]}")
```

### Task 9 (b)

> Transpose the matrix and print its new shape.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 9 (b)
transposed = performance.T

print("Original (3 x 5) - one row per employee:")
print(performance)
print("Shape:", performance.shape)

print("\nTransposed (5 x 3) - one row per month:")
print(transposed)
print("Shape:", transposed.shape)

# Transposing flips rows into columns, so the meaning of each axis swaps too
print("\nJanuary scores for all employees:", transposed[0])
```

### Task 9 (c)

> Verify programmatically (using code, not by manual inspection) that transposing the matrix twice gives back the original matrix.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 9 (c)
double_transposed = performance.T.T

print("Original shape          :", performance.shape)
print("After one transpose     :", performance.T.shape)
print("After two transposes    :", double_transposed.shape)

# np.array_equal compares shape AND every value, returning a single True/False
same = np.array_equal(performance, double_transposed)

print("\nDoes (A.T).T equal A?", same)

if same:
    print("Confirmed - transposing twice returns the original matrix.")
else:
    print("The matrices differ.")
```

### Task 9 (d)

> Create a 4x2 and a 3x2 matrix, and use transpose on one of them so that a dot product between the two becomes possible; then perform the dot product.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 9 (d)
matrix_4x2 = np.array([[1, 2], [3, 4], [5, 6], [7, 8]])
matrix_3x2 = np.array([[2, 0], [1, 3], [4, 5]])

print("matrix_4x2 shape:", matrix_4x2.shape)
print("matrix_3x2 shape:", matrix_3x2.shape)

# (4, 2) . (3, 2) fails because the inner dimensions 2 and 3 do not match
try:
    np.dot(matrix_4x2, matrix_3x2)
except ValueError as err:
    print("\nDirect dot product fails -> ValueError:", err)

# Transposing the second matrix turns (3, 2) into (2, 3), so the inner
# dimensions become 2 and 2 -> the multiplication is now allowed.
result = np.dot(matrix_4x2, matrix_3x2.T)

print("\nAfter transposing: (4, 2) . (2, 3) -> (4, 3)")
print(result)
print("Result shape:", result.shape)
```
